## 2. Module Overview and Design

### File Format Understanding

The input files have a specific structure:
```
Question 1. [Text of question 1]
[x] Answer 1.1
[ ] Answer 1.2
[ ] Answer 1.3
[ ] Answer 1.4
```

Key parsing challenges:
- Identifying question boundaries
- Distinguishing between selected `[x]` and unselected `[ ]` answers
- Handling unanswered questions (all brackets empty)
- Ensuring exactly 100 questions are processed

### Design Decisions

1. **Line-by-line parsing**: Read file sequentially to identify question blocks
2. **State tracking**: Accumulate answer lines until the next question is encountered
3. **Validation**: Verify exactly 100 questions are extracted
4. **Error handling**: Comprehensive exception handling for common issues

### Function 1: `extract_answers_sequence(file_path)`

This function parses a quiz answer file and returns a list of 100 integers.

## 4. Testing and Validation

### Test 1: Creating Sample Data

To test my module, I'll create a sample quiz answer file:

````python
# Create a sample test file with 100 questions
sample_data = []

# Create 100 questions with varying answer patterns
for i in range(1, 101):
    sample_data.append(f"Question {i}. Sample question text for question {i}")
    
    # Determine which answer to mark (create a pattern)
    if i <= 25:
        selected = 1  # First 25 questions: answer 1
    elif i <= 50:
        selected = 2  # Questions 26-50: answer 2
    elif i <= 75:
        selected = 3  # Questions 51-75: answer 3
    elif i <= 95:
        selected = 4  # Questions 76-95: answer 4
    else:
        selected = 0  # Last 5 questions: unanswered
    
    # Add four answer lines
    for ans_num in range(1, 5):
        if ans_num == selected:
            sample_data.append(f"[x] Answer {i}.{ans_num}")
        else:
            sample_data.append(f"[ ] Answer {i}.{ans_num}")

# Write sample file
with open('test_sample.txt', 'w') as f:
    f.write('\n'.join(sample_data))

print("Sample test file created: test_sample.txt")
print(f"Total lines: {len(sample_data)}")
````

### Test 2: Extract Answers from Sample File

````python
# Test the extraction function
try:
    extracted = extract_answers_sequence('test_sample.txt')
    print(f"Successfully extracted {len(extracted)} answers")
    print(f"\nFirst 10 answers: {extracted[:10]}")
    print(f"Answers 21-30: {extracted[20:30]}")
    print(f"Answers 46-55: {extracted[45:55]}")
    print(f"Last 10 answers: {extracted[-10:]}")
    
    # Verify the expected pattern
    print("\n--- Pattern Verification ---")
    print(f"Questions 1-25 (expected all 1s): {set(extracted[0:25])}")
    print(f"Questions 26-50 (expected all 2s): {set(extracted[25:50])}")
    print(f"Questions 51-75 (expected all 3s): {set(extracted[50:75])}")
    print(f"Questions 76-95 (expected all 4s): {set(extracted[75:95])}")
    print(f"Questions 96-100 (expected all 0s): {set(extracted[95:100])}")
    
except Exception as e:
    print(f"Error during extraction: {e}")
````

### Test 3: Write Sequence to File

````python
# Test the write function
try:
    write_answers_sequence(extracted, 999)  # Test respondent ID 999
    
    # Verify the file was created and read it back
    with open('answers_list_respondent_999.txt', 'r') as f:
        written_answers = [int(line.strip()) for line in f.readlines()]
    
    print(f"File created successfully!")
    print(f"Wrote {len(written_answers)} answers")
    print(f"Verification: Original == Written? {extracted == written_answers}")
    
except Exception as e:
    print(f"Error during writing: {e}")
````

### Test 4: Error Handling

````python
# Test 4a: Non-existent file
print("Test 4a: Non-existent file")
try:
    extract_answers_sequence('nonexistent_file.txt')
except FileNotFoundError as e:
    print(f"✓ Correctly caught FileNotFoundError: {e}")

# Test 4b: Invalid answer count
print("\nTest 4b: Invalid answer count")
try:
    write_answers_sequence([1, 2, 3], 1)  # Only 3 answers instead of 100
except ValueError as e:
    print(f"✓ Correctly caught ValueError: {e}")

# Test 4c: Invalid answer values
print("\nTest 4c: Invalid answer values")
try:
    invalid_answers = [1] * 99 + [5]  # Last answer is 5 (out of range)
    write_answers_sequence(invalid_answers, 1)
except ValueError as e:
    print(f"✓ Correctly caught ValueError: {e}")
````

### Potential Analysis Approaches

Once real data is available, I could investigate:

````python
# Example analysis on sample data
import matplotlib.pyplot as plt
import numpy as np

# Using the extracted sample data
if 'extracted' in locals():
    # Remove zeros (unanswered) for some analyses
    answered_only = [x for x in extracted if x != 0]
    
    print("Sample Data Analysis:")
    print(f"Total questions: {len(extracted)}")
    print(f"Answered questions: {len(answered_only)}")
    print(f"Unanswered questions: {extracted.count(0)}")
    print(f"\nAnswer distribution:")
    for ans in [1, 2, 3, 4]:
        count = extracted.count(ans)
        print(f"  Answer {ans}: {count} times ({count/len(extracted)*100:.1f}%)")
    
    # Visualization
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Scatter plot
    ax1.scatter(range(1, 101), extracted, alpha=0.6)
    ax1.set_xlabel('Question Number')
    ax1.set_ylabel('Answer Selected')
    ax1.set_title('Answer Sequence Scatter Plot')
    ax1.set_ylim(-0.5, 4.5)
    ax1.grid(True, alpha=0.3)
    
    # Bar chart of distribution
    answers_count = [extracted.count(i) for i in range(5)]
    ax2.bar(['0\n(Unanswered)', '1', '2', '3', '4'], answers_count, color=['gray', 'blue', 'green', 'orange', 'red'])
    ax2.set_xlabel('Answer')
    ax2.set_ylabel('Frequency')
    ax2.set_title('Answer Distribution')
    ax2.grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    plt.show()
    
    print("\nPattern observation: The sample data shows clear blocks of answers,")
    print("which was expected as I created it with that pattern for testing.")
````
