## Test of my function

Before using the real project file, I tested my function with a small artificial collated file. This helps check that `0` values are ignored correctly.

## Interpretation

After running the real analysis, I would compare the scatter plot and line plot to check whether the mean answer values show a visible structure across question number.

The key points I would look for are:

- whether the mean values stay close to a random-looking spread;
- whether there is a trend across the 100 questions;
- whether there are repeated blocks, cycles, or alternating behaviour;
- whether some questions have unusual mean values compared with nearby questions.

If the plot does not show a clear trend or repeated structure, then the evidence from this part of the analysis would be weak for a deliberate pattern. If the line plot shows regular rises, falls, or repeated blocks, then this could support the idea that there may be a hidden pattern, but it would still need to be checked with other evidence from the group.
